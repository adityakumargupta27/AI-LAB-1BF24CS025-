```python
import heapq

GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)
INITIAL_STATE = (1, 3, 5, 4, 2, 6, 7, 0, 8)

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(list(state[i:i+3]))

def get_neighbors(state):
    neighbors = []
    idx = state.index(0)
    row, col = idx // 3, idx % 3
    moves = []
    if row > 0: moves.append((-1, 0, 'Up'))
    if row < 2: moves.append((1, 0, 'Down'))
    if col > 0: moves.append((0, -1, 'Left'))
    if col < 2: moves.append((0, 1, 'Right'))
    for dr, dc, action in moves:
        new_row, new_col = row + dr, col + dc
        new_idx = new_row * 3 + new_col
        new_state = list(state)
        new_state[idx], new_state[new_idx] = new_state[new_idx], new_state[idx]
        neighbors.append((tuple(new_state), action))
    return neighbors

def h_manhattan(state, goal=GOAL_STATE):
    distance = 0
    for i, val in enumerate(state):
        if val != 0:
            curr_row, curr_col = i // 3, i % 3
            target_idx = goal.index(val)
            target_row, target_col = target_idx // 3, target_idx % 3
            distance += abs(curr_row - target_row) + abs(curr_col - target_col)
    return distance

def a_star_search(start_state, heuristic, goal_state=GOAL_STATE):
    heap = []
    heapq.heappush(heap, (heuristic(start_state, goal_state), 0, start_state, []))
    visited = set()
    nodes_expanded = 0
    while heap:
        fn, gn, current_state, path = heapq.heappop(heap)
        if current_state in visited:
            continue
        visited.add(current_state)
        nodes_expanded += 1
        if current_state == goal_state:
            return path, nodes_expanded
        for neighbor, move in get_neighbors(current_state):
            if neighbor not in visited:
                next_gn = gn + 1
                next_hn = heuristic(neighbor, goal_state)
                next_fn = next_gn + next_hn
                heapq.heappush(heap, (next_fn, next_gn, neighbor, path + [move]))
    return None, nodes_expanded

print("=== Initial State ===")
print_puzzle(INITIAL_STATE)
print("\n=== Goal State ===")
print_puzzle(GOAL_STATE)
print()

path_manhattan, nodes_manhattan = a_star_search(INITIAL_STATE, h_manhattan)
print("--- Manhattan Distance Heuristic ---")
print(f"Moves: {len(path_manhattan) if path_manhattan else 'None'} -> {path_manhattan}")
print(f"Nodes expanded: {nodes_manhattan}")
```
