```python
import heapq

GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)
INITIAL_STATE = (1, 2, 3, 0, 4, 6, 7, 5, 8)

def h(state, goal=GOAL_STATE):
    misplaced = 0
    for s, g_val in zip(state, goal):
        if s != 0 and s != g_val:
            misplaced += 1
    return misplaced

def g(depth):
    return depth

def f(gn, hn):
    return gn + hn

def get_neighbors(state):
    neighbors = []
    blank_idx = state.index(0)
    row, col = blank_idx // 3, blank_idx % 3
    moves = {
        'Up': (-1, 0),
        'Down': (1, 0),
        'Left': (0, -1),
        'Right': (0, 1)
    }
    for direction, (dr, dc) in moves.items():
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank_idx = new_row * 3 + new_col
            state_list = list(state)
            state_list[blank_idx], state_list[new_blank_idx] = state_list[new_blank_idx], state_list[blank_idx]
            neighbors.append((tuple(state_list), direction))
    return neighbors

def a_star_search(start_state, goal_state=GOAL_STATE):
    heap = []
    initial_gn = g(0)
    initial_hn = h(start_state, goal_state)
    initial_fn = f(initial_gn, initial_hn)
    heapq.heappush(heap, (initial_fn, initial_gn, start_state, []))
    visited = set()
    nodes_expanded = 0
    while heap:
        fn, gn, current_state, path = heapq.heappop(heap)
        if current_state in visited:
            continue
        visited.add(current_state)
        nodes_expanded += 1
        if current_state == goal_state:
            return path, nodes_expanded
        for neighbor, move in get_neighbors(current_state):
            if neighbor not in visited:
                next_gn = g(gn + 1)
                next_hn = h(neighbor, goal_state)
                next_fn = f(next_gn, next_hn)
                heapq.heappush(heap, (next_fn, next_gn, neighbor, path + [move]))
    return None, nodes_expanded

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(list(state[i:i+3]))
    print()

def validate_state(state):
    if not isinstance(state, tuple) or len(state) != 9:
        return False
    return sorted(list(state)) == list(range(9))

if not validate_state(INITIAL_STATE) or not validate_state(GOAL_STATE):
    print("Error: States must be tuples containing unique integers from 0 to 8.")
else:
    print("Initial State:")
    print_puzzle(INITIAL_STATE)
    print("Goal State:")
    print_puzzle(GOAL_STATE)

    path, nodes = a_star_search(INITIAL_STATE, GOAL_STATE)
    if path is not None:
        print(f"Solution found in {len(path)} moves.")
        print("Moves:", " -> ".join(path))
        print(f"Nodes expanded: {nodes}")
    else:
        print("No solution found.")
```
