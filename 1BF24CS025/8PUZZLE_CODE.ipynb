```python
initial_board = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

goal_board = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

initial_state = PuzzleState(initial_board)
goal_state = PuzzleState(goal_board)

print("Initial State:")
print(initial_state)
print("\nGoal State:")
print(goal_state)

print("\nSolving using DFS...")
solution = dfs_solve(initial_state, goal_state)

if solution:
    print("\nSolution Found!")
    path = reconstruct_path(solution)
    print(f"Number of moves: {len(path)}")
    print(f"Path: {path}")

    # Display the steps
    current = initial_state
    print("\nSteps:")
    print("Initial State:")
    print(current)
    for i, action in enumerate(path):
        print(f"\nStep {i+1}: {action}")
        for neighbor in current.get_neighbors():
            if neighbor.action == action:
                current = neighbor
                break
        print(current)
else:
    print("\nNo solution found for this initial state (or DFS couldn't find it within practical limits).")
```
